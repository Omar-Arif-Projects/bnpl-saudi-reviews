# Gold-set labeling

Hand labels for 400 reviews, used as the test set for the sentiment and topic models. The sample is stratified by app, language (Arabic, English) and star group, with negative and neutral reviews over-sampled. It is created with:

```bash
python -m src.label.gold --sample
```

Each label is saved to `data/labeled/gold.csv` immediately, so labeling can stop and resume at any time. Star ratings are hidden while labeling.

## Labeling guide

**Sentiment** is judged from the text alone.

| Label | Use when |
|---|---|
| Negative | The review mainly complains, warns others or expresses dissatisfaction. |
| Neutral | No clear opinion: a question, a factual statement, a feature request without complaint, or praise and complaint in equal measure. |
| Positive | The review mainly praises, thanks or recommends. |

- Mixed reviews take the dominant sentiment ("good app but the late fee is unfair" is negative if the complaint is the point).
- Sarcasm takes the intended meaning ("شكرا على رفض طلبي 👏" is negative).
- A request phrased as a complaint ("why is my limit so low?") is negative.

**Main topic** is what the opinion is about. Choose one; if several apply, choose the one the review spends most words on. Short generic reviews ("ممتاز", "good") are *General opinion*.

**Flags** exclude a row from evaluation:
- *Unclear*: the meaning or sentiment cannot be determined.
- *Not about app*: spam, or about something unrelated.
- *Personal info*: contains a name, phone number or other personal detail; its text is removed from the file.

## Blind relabel

After all rows are labeled, draw a random set of labeled rows (`gold.relabel_n` in `config.yaml`) with their labels blanked:

```bash
python -m src.label.gold --relabel
```

The rows are saved to `data/labeled/gold_relabel.csv`. Label them below without looking at the first labels. `python -m src.models.evaluate` then compares the two labelings and writes Cohen's kappa to `reports/gold_agreement.csv`.
